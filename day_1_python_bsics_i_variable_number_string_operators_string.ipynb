# **Basic Operators**

1. **Arithmetic operators:** +, -, *, / etc.

2. **Assignment operators:** =, +=, -= etc.

3. **Comparison operators:** ==, >, >=, <, != etc.

4. **Logical operators:** and, or, not.









```python
a = 11
b = 3

# Addition
print("Addition:", a + b)

# Subtraction
print("Subtraction:", a - b)

# Multiplication
print("Multiplication:", a * b)

# Normal Division (decimal result)
print("Division:", a / b)

# Floor Division (integer result, no decimal)
print("Floor Division:", a // b)

# Modulus (remainder)
print("Modulus:", a % b)
```

```python
#Assignment operators
x = 5
print("x =", x)
x=5
x=x+5 #x=10
print(x)
# Add and assign
x += 5   # x = x + 5
print("x after += 5:", x)

# Subtract and assign
x -= 3   # x = x - 3
print("x after -= 3:", x)

# Multiply and assign
x *= 2   # x = x * 2
print("x after *= 2:", x)
```

```python
#Comparison operators
a = 10
b = 5

print("a == b:", a == b)   # equal
print("a != b:", a != b)   # not equal
print("a > b:", a > b)     # greater than
print("a < b:", a < b)     # less than
print("a >= b:", a >= b)   # greater or equal
print("a <= b:", a <= b)   # less or equal
```

```python
#Logical operators
x = True
y = False

print("x and y:", x and y)   # both True
print("x or y:", x or y)     # at least one True
print("not x:", not x)       # reverses the value
```
